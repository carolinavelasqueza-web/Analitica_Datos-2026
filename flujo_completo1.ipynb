```python
# Cargar variables del archivo .env
from dotenv import load_dotenv
load_dotenv()

import os
from hf_dataset_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer

# Verificar que el token esté cargado
token = os.environ.get("HF_TOKEN")
if token:
    print(f"✅ Token configurado: {token[:10]}...")
else:
    print(" ERROR: HF_TOKEN no está configurado. Revisa el archivo .env")
```

```python
# Usamos scikit-learn/iris
loader = HFDatasetLoader("scikit-learn/iris", split="train")
df = loader.load()

print(f"✅ Dataset cargado: {df.shape[0]} filas, {df.shape[1]} columnas")
print(f"Columnas originales: {list(df.columns)}")
df.head(10)
```

```python
# Renombrar columnas 
df = df.rename(columns={
    "SepalLengthCm": "longitud_sepalo",
    "SepalWidthCm": "ancho_sepalo",
    "PetalLengthCm": "longitud_petalo",
    "PetalWidthCm": "ancho_petalo",
    "Species": "especie"
})

print("✅ Transformación 1: Renombrar columnas")
print(f"Columnas después: {list(df.columns)}")
df.head()
```

```python
# Crear columna que calcula el ratio entre longitud y ancho del sépalo
df["ratio_sepalo"] = df["longitud_sepalo"] / df["ancho_sepalo"]

print("✅ Transformación 2: Crear columna 'ratio_sepalo'")
print(f"Estadísticas de ratio_sepalo:\n{df['ratio_sepalo'].describe()}")
df[["longitud_sepalo", "ancho_sepalo", "ratio_sepalo"]].head(10)
```

```python
# Filtrar solo flores con ratio_sepalo mayor a 1.5
df_filtrado = df[df["ratio_sepalo"] > 1.5].copy()

print(f"✅ Transformación 3: Filtrar ratio_sepalo > 1.5")
print(f"Filas antes: {df.shape[0]} → Filas después: {df_filtrado.shape[0]}")
df_filtrado.head()
```

```python
# Seleccionar solo las columnas más relevantes
columnas_relevantes = [
    "longitud_sepalo", "ancho_sepalo", "ratio_sepalo", "especie"
]

df_final = df_filtrado[columnas_relevantes].copy()

print("✅ Transformación 4: Seleccionar columnas relevantes")
print(f"Columnas finales: {list(df_final.columns)}")
print(f"Dataset final: {df_final.shape[0]} filas, {df_final.shape[1]} columnas")
df_final.head(10)
```

```python
summarizer = DatasetSummarizer(df_final, title="Resumen Dataset Iris - Transformado")
ruta = summarizer.save_txt("resumen_dataset.txt")
print(f"✅ Resumen guardado en: {ruta}")
```
